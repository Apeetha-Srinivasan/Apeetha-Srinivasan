```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv(r"C:\Users\apisr\Downloads\Teen_Mental_Health_Dataset.csv")
```

```python
#categorical data count plot
# bar x-axis categorical  y-axis numerical
# corelation scatter plot
#line chart - analyse trend, countinous data, or time
# box plot - to see ho data is spread, outlier
#hist frequency of each data
#density - distribution of data, smoother version of hist
#violin box + density
#correlation heatmap= correlation
#pairplot=
# plt.figure(figsize=(8,5))
```

```python
plt.figure(figsize = (3,3))
sns.countplot(x = 'gender', data = df)
plt.title("Count of Gender")
plt.show()
```

```python
plt.figure(figsize=(3,3))
sns.barplot(x="gender",y="stress_level",data =df)
plt.title("Average Stress Level by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Stress Level")
plt.show()
```

```python
plt.figure(figsize=(6,3))
sns.scatterplot(x = "sleep_hours", y = "stress_level",data = df)
plt.title("Study Hours Vs Stress Level")
plt.xlabel("Study Hours")
plt.ylabel("Stress Level" )
```

```python
plt.figure(figsize=(5,3))
sns.lineplot(x = "sleep_hours", y = "stress_level",data = df)
plt.title("Study Hours Vs Stress Level")
plt.xlabel("Study Hours")
plt.ylabel("Stress Level")
```

```python
plt.figure(figsize=(5,3))
sns.boxplot(x="gender", y = "anxiety_level", data = df)
plt.title("Anxiety Level by Gender")
plt.xlabel("Gender")
plt.ylabel("Anxiety Level")
plt.show()
```

```python
sns.kdeplot(df["stress_level"], fill = True)
plt.title("Distribution of Stress Level")
plt.xlabel("Stress Level")
plt.ylabel("Density")
plt.show()
```

```python
sns.histplot(df["sleep_hours"], bins = 15)
plt.title("Distribution of Sleep Hours")
plt.xlabel("Sleep Hours")
plt.ylabel("Frequency")
plt.show()
```

```python
#hist with 2 vayes x and y
sns.histplot(x = df["sleep_hours"], y = df["stress_level"], bins = 15, cbar = True )
plt.title("Distribution of Sleep Hours")
plt.xlabel("Sleep Hours")
plt.ylabel("Stress Level")
plt.show()
```

```python
sns.violinplot(x="gender", y ="anxiety_level", data = df)
plt.title("Anxiety Level by Gender")
plt.xlabel("Gender")
plt.ylabel("Anxiety Level")
plt.show()
```

```python
sns.swarmplot(x= "gender", y = "stress_level", data = df)
plt.title("Stress Level by Gender")
plt.xlabel("Gender")
plt.ylabel("Stress Level")
plt.show()
```

```python
plt.figure(figsize = (8,5))
numeric_df = df.select_dtypes(include=['int64','float64'])
sns.heatmap(numeric_df.corr(), annot = True, cmap = 'coolwarm')
plt.title("Correlation Heatmap")
plt.show()
```

```python
sns.pairplot(numeric_df)
plt.show()
```
