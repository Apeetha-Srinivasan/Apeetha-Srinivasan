```python
#while loop with loop control statements

total_bill_amount=0
i=1
print("Enter the product prices one by one: ")
while True:
    price=float(input())
    if price<=0:
        break
    total_bill_amount += price
    i+=1
print("Total bill amount",total_bill_amount)
```
